```python
import os
import matplotlib.pyplot as plt
import numpy as np

# the path of images
pasta = r"C:\Users\luisf\Documents\brain-tumor-detection-yolov11\brain-tumor-detection-yolov11\data\train\images"
caminho = [
    os.path.join(pasta, f) 
    for f in os.listdir(pasta) 
    if f.lower().endswith(".jpg")
]

# starting the subplots
fig, axs = plt.subplots(6, 6)

# using lambda to plot the images on the subplots and define the axis for each image at path
list(map(lambda ax, cam: ax.imshow(plt.imread(cam)), axs.flat, caminho))

list(map(lambda ax: ax.axis('off'), axs.flat))

# plotting the image
plt.tight_layout() # -> ajusta os subplots
plt.show()
```
